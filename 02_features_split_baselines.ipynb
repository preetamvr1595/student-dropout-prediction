## 3. Train / test split

We hold out **20%** as a test set, **stratified** so both parts keep the same dropout rate. The test set is used **once**, at the end of notebook 03. All tuning uses cross-validation on the training part only.

## 4. Metrics ΓÇö what matters here?

| Metric | Why it matters for dropout |
|---|---|
| **Recall** | of students who *will* drop out, how many do we catch? (missing one = a student not helped) |
| **Precision** | of students we flag, how many really are at risk? (wasted counselling effort) |
| F1 | balance of the two |
| ROC-AUC | ranking quality over all thresholds |
| **PR-AUC** | better than ROC-AUC when the positive class (dropout) is the minority |

Accuracy is reported but not trusted alone: predicting *nobody drops out* already gets Γëê 68%.

```python
def metrics(y_true, proba, thr=.5):
    pred=(proba>=thr).astype(int)
    return dict(accuracy=accuracy_score(y_true,pred), precision=precision_score(y_true,pred,zero_division=0), recall=recall_score(y_true,pred), f1=f1_score(y_true,pred), roc_auc=roc_auc_score(y_true,proba), pr_auc=average_precision_score(y_true,proba))
metrics(y[idx_te], np.zeros(len(idx_te)))
```

The ΓÇ£nobody drops outΓÇ¥ rule: accuracy Γëê 0.68 but recall 0 and ROC-AUC 0.5 ΓÇö exactly why we need the other metrics.

## 5. Baseline models with cross-validation

```python
cat_cols=[c for c in CATEGORICAL if c in X1.columns]
def make_prep(X):
    cats=[c for c in CATEGORICAL if c in X.columns]; nums=[c for c in X.columns if c not in cats]
    return ColumnTransformer([('cat',OneHotEncoder(handle_unknown='ignore',min_frequency=10),cats),('num',StandardScaler(),nums)])
make_prep(X1)
```

```python
skf = StratifiedKFold(5, shuffle=True, random_state=SEED)
SCORING={'roc_auc':'roc_auc','pr_auc':'average_precision','f1':'f1','recall':'recall','precision':'precision','accuracy':'accuracy'}
def cv_model(name, model, X):
    pipe=Pipeline([('prep',make_prep(X)),('model',model)])
    t=time.time(); r=cross_validate(pipe, X.loc[idx_tr], y[idx_tr], cv=skf, scoring=SCORING, n_jobs=2)
    return {'model':name, **{k:r['test_'+k].mean() for k in SCORING}, **{k+'_std':r['test_'+k].std() for k in ['roc_auc']}, 'sec':time.time()-t}
ROWS=[]
```

### 5.1 Dummy (majority class)

```python
ROWS.append({'stage':'sem1', **cv_model('Dummy (majority)', DummyClassifier(strategy='most_frequent'), X1)}); pd.DataFrame(ROWS).round(3)
```

### 5.2 Logistic regression

```python
ROWS.append({'stage':'sem1', **cv_model('Logistic Regression', LogisticRegression(max_iter=2000,C=0.5), X1)}); pd.DataFrame(ROWS).round(3)
```

```python
ROWS.append({'stage':'sem1', **cv_model('Logistic Regression (balanced)', LogisticRegression(max_iter=2000,C=0.5,class_weight='balanced'), X1)}); pd.DataFrame(ROWS).round(3).tail(1)
```

### 5.3 Decision tree

```python
ROWS.append({'stage':'sem1', **cv_model('Decision Tree (depth 5)', DecisionTreeClassifier(max_depth=5,class_weight='balanced',random_state=SEED), X1)}); pd.DataFrame(ROWS).round(3).tail(1)
```

### 5.4 Random forest

```python
ROWS.append({'stage':'sem1', **cv_model('Random Forest', RandomForestClassifier(n_estimators=300,min_samples_leaf=3,class_weight='balanced_subsample',n_jobs=2,random_state=SEED), X1)}); pd.DataFrame(ROWS).round(3).tail(1)
```

## 6. The same baselines at stage 2 (after semester 2)

```python
for name,model in [('Logistic Regression (balanced)', LogisticRegression(max_iter=2000,C=0.5,class_weight='balanced')),('Random Forest', RandomForestClassifier(n_estimators=300,min_samples_leaf=3,class_weight='balanced_subsample',n_jobs=2,random_state=SEED))]:
    ROWS.append({'stage':'sem2', **cv_model(name, model, X2)})
pd.DataFrame(ROWS).round(3)
```

```python
res=pd.DataFrame(ROWS); res['label']=res.stage+' ┬╖ '+res.model
res.set_index('label')[['roc_auc','pr_auc','f1','recall','precision','accuracy']].round(3).sort_values('roc_auc',ascending=False)
```

```python
r=res.set_index('label')['roc_auc'].sort_values(); r.plot(kind='barh',figsize=(8,4),color='steelblue'); plt.xlim(.45,1); plt.xlabel('CV ROC-AUC'); plt.title('Baselines: sem1 (early) vs sem2 (later)'); plt.tight_layout(); plt.show()
```

Notice: (1) the dummy model scores ROC-AUC 0.5; (2) all real models are far above it; (3) stage 2 is better than stage 1 ΓÇö more information ΓÇö but arrives a whole semester later.

## 7. Where does a simple model go wrong?

```python
pipe=Pipeline([('prep',make_prep(X1)),('model',LogisticRegression(max_iter=2000,C=0.5,class_weight='balanced'))]).fit(X1.loc[idx_tr],y[idx_tr])
proba=pipe.predict_proba(X1.loc[idx_te])[:,1]; pd.Series(metrics(y[idx_te],proba)).round(3).to_frame('baseline LR (test set, illustration only)')
```

*(We peek at the test set here only for an illustration; model choices in notebook 03 use CV on train.)*

```python
ConfusionMatrixDisplay.from_predictions(y[idx_te],(proba>=.5).astype(int),display_labels=['stay','dropout'],cmap='Blues'); plt.title('Baseline LR ΓÇö confusion matrix'); plt.show()
```

```python
fpr,tpr,_=roc_curve(y[idx_te],proba); pr,rc,_=precision_recall_curve(y[idx_te],proba)
fig,ax=plt.subplots(1,2,figsize=(11,4)); ax[0].plot(fpr,tpr); ax[0].plot([0,1],[0,1],'k--'); ax[0].set_title('ROC'); ax[1].plot(rc,pr); ax[1].axhline(y[idx_te].mean(),color='k',ls='--'); ax[1].set_title('Precision-Recall'); ax[1].set_xlabel('recall'); plt.show()
```
